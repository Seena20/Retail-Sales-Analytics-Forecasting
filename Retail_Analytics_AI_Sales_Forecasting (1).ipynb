Forecasting Dataset Prepare

```python
daily_sales = daily_sales.sort_values('Date').reset_index(drop=True)

daily_sales.head()
```

## Train-Test Split

The daily sales data was divided chronologically into training and testing datasets.

- Training Data: 641 days
- Testing Data: 90 days

The last 90 days were reserved for testing the forecasting model on unseen data.

```python
train = daily_sales.iloc[:-90].copy()
test = daily_sales.iloc[-90:].copy()

print("Training Data:", len(train))
print("Testing Data:", len(test))
```

```python
test['Baseline_Prediction'] = train['Total_Sales'].iloc[-1]
```

## Feature Engineering

Historical sales features were created to help the Machine Learning models learn recent sales patterns.

The main features include Lag 1 Day, Lag 7 Days, Lag 30 Days, 7-Day Rolling Average, and 30-Day Rolling Average.

```python
forecast_df = daily_sales.copy()

forecast_df['Day'] = forecast_df['Date'].dt.day
forecast_df['Month'] = forecast_df['Date'].dt.month
forecast_df['Year'] = forecast_df['Date'].dt.year
forecast_df['DayOfWeek'] = forecast_df['Date'].dt.dayofweek

forecast_df['Lag_1'] = forecast_df['Total_Sales'].shift(1)
forecast_df['Lag_7'] = forecast_df['Total_Sales'].shift(7)
forecast_df['Lag_30'] = forecast_df['Total_Sales'].shift(30)

forecast_df['Rolling_7'] = forecast_df['Total_Sales'].rolling(7).mean()
forecast_df['Rolling_30'] = forecast_df['Total_Sales'].rolling(30).mean()

forecast_df.head(10)
```

```python
forecast_df = forecast_df.dropna().reset_index(drop=True)

forecast_df.shape
```

#  Machine Learning Model Development

Machine Learning models were developed to forecast daily retail sales.

The following models were evaluated:

1. Random Forest Regressor
2. Gradient Boosting Regressor
3. Improved Gradient Boosting Regressor

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
features = [
    'Day',
    'Month',
    'Year',
    'DayOfWeek',
    'Lag_1',
    'Lag_7',
    'Lag_30',
    'Rolling_7',
    'Rolling_30'
]

X = forecast_df[features]
y = forecast_df['Total_Sales']
```

```python
X_train = X.iloc[:-90]
X_test = X.iloc[-90:]

y_train = y.iloc[:-90]
y_test = y.iloc[-90:]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred[:10])
```

##  Model Evaluation

The models were evaluated using three performance metrics:

- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)
- R² Score

Lower MAE and RMSE indicate lower prediction errors, while a higher R² score indicates better explanatory performance.

```python
mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
plt.figure(figsize=(14,6))

plt.plot(test['Date'], y_test.values, label='Actual Sales')
plt.plot(test['Date'], y_pred, label='Predicted Sales')

plt.title('Actual vs Predicted Sales - Random Forest')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
feature_importance = pd.DataFrame({
    'Feature': features,
    'Importance': model.feature_importances_
}).sort_values('Importance', ascending=False)

print(feature_importance)
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    feature_importance['Feature'],
    feature_importance['Importance']
)

plt.title('Feature Importance - Random Forest')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.gca().invert_yaxis()

plt.show()
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

gb_model.fit(X_train, y_train)

gb_pred = gb_model.predict(X_test)
```

```python
gb_mae = mean_absolute_error(y_test, gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))
gb_r2 = r2_score(y_test, gb_pred)

print("Gradient Boosting Results")
print("-------------------------")
print("MAE :", gb_mae)
print("RMSE:", gb_rmse)
print("R² Score:", gb_r2)
```

```python
future_days = 30

last_date = daily_sales['Date'].max()

future_dates = pd.date_range(
    start=last_date + pd.Timedelta(days=1),
    periods=future_days,
    freq='D'
)

future_dates
```

```python
future_df = pd.DataFrame({
    'Date': future_dates
})

future_df['Day'] = future_df['Date'].dt.day
future_df['Month'] = future_df['Date'].dt.month
future_df['Year'] = future_df['Date'].dt.year
future_df['DayOfWeek'] = future_df['Date'].dt.dayofweek
```

```python
# Historical sales values ko list mein store karna
history = daily_sales['Total_Sales'].tolist()

future_predictions = []

for date in future_dates:

    # Future date ke features
    day = date.day
    month = date.month
    year = date.year
    day_of_week = date.dayofweek

    # Lag features
    lag_1 = history[-1]
    lag_7 = history[-7]
    lag_30 = history[-30]

    # Rolling averages
    rolling_7 = np.mean(history[-7:])
    rolling_30 = np.mean(history[-30:])

    # Model input
    input_data = pd.DataFrame([[
        day,
        month,
        year,
        day_of_week,
        lag_1,
        lag_7,
        lag_30,
        rolling_7,
        rolling_30
    ]], columns=features)

    # Prediction
    prediction = gb_model.predict(input_data)[0]

    # Store prediction
    future_predictions.append(prediction)

    # Prediction ko history mein add karna
    history.append(prediction)
```
